```python
ch = 65       
row = 1  # 2 3 4

while row <= 7:   # T T T  
    col = 1   

    while col <= row:  # T T T F
        print(chr(ch), end=" ")
        if chr(ch)=="Z":
            break
        ch += 1  # 66 67 68 69 70
        col += 1  # 2 3 4

    print()
    row += 1
```

```python
row = 1

while row <= 5:

    space = 5 - row
    while space > 0:
        print(" ", end="")
        space -= 1

    star = 1
    while star <= (2 * row - 1):
        print("*", end="")
        star += 1

    print()
    row += 1
```
