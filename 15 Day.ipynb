```python
try:
    with open ("new.txt", "r") as f:
        print(f.read())
        f.close()
except Exception as e:
    print(e)
```

```python
try:
    num = int("42")
    print(f"Converted: {num}")
except:
    print("Conversion failed.")
else:
    print("else block: no error occurred!")
    print(f"Double: {num * 2 }")
finally:
    print ("finally: this ALWAYS runs (cleanup code goes here)")
    
```

```python
print("--- File read demo ---")
try:
    f = open("sample.txt", "r")
    content = f.read()
    print(f"File read successfully. Length: {len(content)} chars")
except FileNotFoundError:
    print("File not found.")
else:
    print("else: file read without any error.")
finally:
    print("finally: this ALWAYS runs (cleanup code goes here)")
```

# logging

```python
# Level hierarchy (Low to high):
# DEBUG     (10) -> very detailed, used while developing/debugging 
# INFO      (20) -> general info about what the program is doing
# WARNING   (30) -> something unexpected but program continues (DEFAULT Level)
# ERROR     (40) -> a real error, something went wrong
# CRITICAL  (50) -> very serious, program might stop working
```

```python
import logging
```

```python
logging.basicConfig(level=logging.DEBUG,                 # show DEBUG and above
    format="%(asctime)s - %(levelname)s - %(message)s")
```

```python
logging.basicConfig(
    filename = "app.log",
    filemode = "a",                                     # append mode
    level = logging.DEBUG,
    format = "%(asctime)s - %(levelname)s - %(message)s",
    datefmt = "%Y - %m - %d %H:%M:%S"
)
```

```python
logging.debug("welcome!")
```

```python
try:
    logging.info("code started")
    with open ("new.txt", "r") as f:
        print(f.read())
        f.close()
except Exception as e:
    print(e)
    logging.error(e)
```

```python
try: 
    logging.info("loop started")
    for i in range(4):
        logging.info(f"variable value {i}")
        print(i)
except Exception as e:
    print(e)
    logging.warning(e)
    logging.error("except block")
```

```python
r = 5
for i in range(1, r+1):
    for j in range (r-i):
        print(" ", end = " ")
    for j in range(i):
        print("*", end = " ")
    print()
```
