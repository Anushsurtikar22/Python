```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance  # name mangling

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited successfully")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Amount withdrawn successfully")
        else:
            print("Insufficient balance")

    def check_balance(self):
        print("Current balance:", self.__balance)
```

```python
bank = BankAccount(1000)
```

```python
bank.check_balance()
```

```python
bank.deposit(9000)
```

```python
bank.check_balance()
```

```python
bank.withdraw(4000)
```

```python
bank.check_balance()
```

```python
bank.withdraw(8000)
```

```python
class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit successful")
        else:
            print("Invalid amount")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawal successful")
        else:
            print("Invalid withdrawal")

    def get_balance(self):
        return self.__balance
```

```python
bank = BankAccount("rohit", 10000)
```

```python
bank.name
```

```python
bank.get_balance()
```

```python
bank.withdraw(-8800)
```

```python
bank.deposit(10000)
```

```python
bank.get_balance()
```

```python
class PhonePe:
    def pay(self):
        print("Payment made through PhonePe.")

class GooglePay:
    def pay(self):
        print("Payment made through Google Pay.")

class Paytm:
    def pay(self):
        print("Payment made through Paytm.")
```

```python
apps = [PhonePe(), GooglePay(), Paytm()]
```

```python
for app in apps:
    if app == PhonePe():
        print(app.pay())
```

```python
for app in apps:
    app.pay()
```

### Duck typing

```python
class Dog:
    def speak(self):
        print("Dog says Woof")


class Cat:
    def speak(self):
        print("Cat says Meow")
```

```python
def make_sound(animal):
    animal.speak()
```

```python
make_sound(Dog())
```

```python
make_sound(Cat())
```
