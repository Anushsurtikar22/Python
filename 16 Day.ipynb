```python
from datetime import time

# Create specific time
meeting_time = time(15, 11, 0)  #3:11 PM
print(meeting_time)   # Output: 15:11:00

# Access time components
print(f"Hour: {meeting_time.hour}")
print(f"Minute: {meeting_time.minute}")
print(f"Second: {meeting_time.second}")
```

```python
from datetime import datetime

# Get current datetime 
now = datetime.now()
print(now)   # Output: 2026-09-01 15:54:53.679875

# Get only time component
current_time = now.time()
print(current_time)  # Output: 15:54:53.679875

# Microseconds useful for precise measurements
print(f"Microseconds: {current_time.microsecond}")
```

```python
from datetime import datetime, timedelta

#Create datetime with specific time
start_time = datetime(2024, 12, 22, 7, 0, 0)  #7:00 AM
print(f"start: {start_time}")

# Add 2 hours 30 minutes
end_time = start_time + timedelta(hours=2, minutes=30)
print(f"End: {end_time}")

# Calculate duration 
duration = end_time - start_time
print(f"Session duration: {duration}")  # Output: 2:30:00
```
